```python
class emplye:
    def __init__(self, data):
        self.name = data["name"]
        self.id = data["id"]
        self.next = None
        self.pre = None

    def set_id(self, id):
        self.id = id

    def set_name(self, name):
        self.name = name

    def set_(self, data):
        self.name = data["name"]
        self.id = data["id"]


class linkedlist:
    def __init__(self):
        self.h = None
        self.t = None

    def add(self, data):
        nd = emplye(data)
        print(nd.id)
        print("Employee Created")
        if self.h is None:
            self.h = nd
            self.t = nd
        else:
            self.t.next = nd
            nd.pre = self.t  
            self.t = nd
        self.t.next = None

    def search_name(self, name):
        nd = self.h
        while nd is not None and nd.name != name:
            nd = nd.next
        return nd

    def search_id(self, id):
        nd = self.h
        while nd is not None and nd.id != id:
            nd = nd.next
        return nd

    def insert(self, data, loc):
        nd = emplye(data)
        if self.h is None:
            self.h = nd
            self.t = nd
        else:
            if loc == 0:
                nd.next = self.h
                self.h.pre = nd
                self.h = nd
            else:
                i = 0
                cur = self.h
                while cur is not None and i < loc - 1:
                    cur = cur.next
                    i += 1
                if cur is None or cur == self.t:
                    nd.pre = self.t
                    self.t.next = nd
                    self.t = nd
                else:
                    cur.next.pre = nd
                    nd.next = cur.next
                    cur.next = nd
                    nd.pre = cur
        self.t.next = None

    def dis(self):
        nd = self.h
        while nd is not None:
            print(f"The name is: {nd.name} and id is: {nd.id}")
            nd = nd.next

    def dele(self):
        nd = self.h
        while nd is not None:
            temp = nd
            nd = nd.next
            del temp
        self.h = None  
        self.t = None  

    def update(self, data, loc):
        cur = self.h
        i = 0
        while cur is not None and i < loc:
            cur = cur.next
            i += 1
        if cur is None:
            print("Wrong location")
        else:
            cur.name = data["name"]
            cur.id = data["id"]





    def del_eelement(self, loc):
        cur = self.h
        i = 0
        while cur is not None and i < loc:
            cur = cur.next
            i += 1
        if cur is None:
            print("Wrong location")



        else:
            print(f"Deleting: {cur.name}")

            if cur == self.h:
                self.h = cur.next
                if self.h:
                    self.h.pre = None
            

            elif cur == self.t:
                self.t = cur.pre
                if self.t:
                    self.t.next = None

                    
            else:
                if cur.next:
                    cur.next.pre = cur.pre
                if cur.pre:
                    cur.pre.next = cur.next
            del cur



a = linkedlist()
a.add({"name": "Ahmed", "id": 1})
a.add({"name": "Sara", "id": 2})
a.add({"name": "Ali", "id": 3})

```
