```python
class Item:
    def __init__(self, value, weight):
        self.value = value
        self.weight = weight
        self.profit = value / weight  

class Knapsack:
    def __init__(self, max_weight=5):
        self.max_weight = max_weight
        self.items = []  
    
    def insert(self, items_list):
        items_list.sort(key=lambda x: x.profit, reverse=True)
        total_weight = 0

        for item in items_list:
            if total_weight + item.weight <= self.max_weight:
                self.items.append(item)  
                total_weight += item.weight
            else:
                if total_weight<self.max_weight:
                    self.items.append(Item(item.value/ (self.max_weight - total_weight), item.weight / (self.max_weight - total_weight)))
                break  
    
    def display(self):
        for item in self.items:
            print(f"Value: {item.value}, Weight: {item.weight}, Profit: {item.profit:.2f}")


items = [Item(10, 2), Item(15, 3), Item(7, 1)]
knapsack = Knapsack(max_weight=5)
knapsack.insert(items)

```

```python
knapsack.display()
```

```python


class Job:
    def __init__(self, name, deadline, profit):
        self.name = name
        self.deadline = deadline
        self.profit = profit

class JobSelect:
    def __init__(self):
        self.selected_jobs = []  
    
    def schedule_jobs(self, jobs):
        jobs.sort(key=lambda x: x.profit, reverse=True)
        
        max_deadline = max(job.deadline for job in jobs)
        time_slots = [False] * max_deadline  
        
        for job in jobs:
            for t in range(min(job.deadline, max_deadline) - 1, -1, -1):
                if not time_slots[t]:
                    time_slots[t] = True
                    self.selected_jobs.append(job)
                    break
    
    def display_schedule(self):
        for job in self.selected_jobs:
            print(f"Job Name: {job.name}, Deadline: {job.deadline}, Profit: {job.profit}")


jobs = [
    Job("jop1", 1, 100),
    Job("jop2", 1, 10),
    Job("jop3", 2, 100),
    Job("jop4", 2, 200),
    Job("jop5", 3, 100),
]

scheduler = JobSelect()
scheduler.schedule_jobs(jobs)
scheduler.display_schedule()
```

```python
def knapsack_01(objects, capacity):
    profit_row = [0] * (capacity + 1)
    
    for obj in objects:
        for weight in range(capacity, obj["weight"] - 1, -1):  
            profit_row[weight] = max(profit_row[weight], obj["Profit"] + profit_row[weight - obj["weight"]])

    collected_Profit = []
    cap = capacity
    for obj in reversed(objects):  
        if cap >= obj["weight"] and profit_row[cap] == obj["Profit"] + profit_row[cap - obj["weight"]]:
            collected_Profit.append(obj)
            cap -= obj["weight"]

    return collected_Profit
```

```python
def ch_mat(arr):
    n = len(arr) - 1
    table = [[float('inf')] * n for _ in range(n)]
    # inf inf inf inf
    # inf inf inf inf
    # inf inf inf inf
    # inf inf inf inf
    for i in range(n):
        table[i][i] = 0
    # 0   inf inf inf
    # inf 0   inf inf
    # inf inf 0   inf
    # inf inf inf 0

    for length in range(1, n):
        # length from 1 to n
        for i in range(n - length):
            # from 0 to  n - length
            j = i + length
            # j = range(n)
            for k in range(i , j):
                # n - length to n
                cost = table[i][k] + table[k+1][j] + arr[i] * arr[k+1] * arr[j+1]
                table[i][j] = min(table[i][j], cost)
                

    return table[0][n - 1]
```
