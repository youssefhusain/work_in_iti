```python
class Stack:
    def __init__(self, size):
        self.stack = []
        self.size = size

    def push(self, item):
        if len(self.stack) < self.size:
            self.stack.append(item)

    def pop(self):
        if self.stack:
            return self.stack.pop()
        return None

numper = Stack(50)
oper = Stack(50)

dic_priority = {
    "-": 0,
    "+": 0,
    "*": 1,
    "/": 1,
    "(": 2,
    ")": 2
}

def apply_operation(a, b, op):
    if op == '+': return a + b
    if op == '-': return a - b
    if op == '*': return a * b
    if op == '/': return a / b

data = input()

i = 0
while i < len(data):
    char = data[i]

    if char.isdigit():
        num = 0
        while i < len(data) and data[i].isdigit():
            num = num * 10 + int(data[i])
            i += 1
        numper.push(num)
        i -= 1
    
    elif char in dic_priority:
        if char == '(':
            oper.push(char)
        elif char == ')':
            while True:
                op = oper.pop()
                if op == '(':
                    break
                b = numper.pop()
                a = numper.pop()
                numper.push(apply_operation(a, b, op))


                
        else:
            while True:
                if not oper.stack:
                    break
                top_op = oper.pop()
                if top_op == '(' or dic_priority[top_op] < dic_priority[char]:
                    oper.push(top_op)
                    break
                b = numper.pop()
                a = numper.pop()
                numper.push(apply_operation(a, b, top_op))
            oper.push(char)
    
    i += 1

while oper.stack:
    op = oper.pop()
    b = numper.pop()
    a = numper.pop()
    numper.push(apply_operation(a, b, op))
print(f"the output of {data} = {numper.pop()}")
```

```python
class queue:
    def __init__(self,س=5):
       self.arr = []
       self.out=0
       self.size = س

    def enqueue(self,data):
        if self.out < self.size:
            self.arr.append(data)
            self.out=len(self.arr)
            print("data is add to queue")
        else :
            print("queue is full")
    def dequeue(self):
        if self.out > 0:
            temp = self.arr[0]
            self.arr=self.arr[1:]
            self.out=len(self.arr)
            print(f"data is remove  from  queue is {temp}")
        else :
            print("queue is empty")
a= queue(5)
```
