```python
import pandas
import matplotlib.pyplot as plt
```

## Loading dataset

First, we use pandas to load the dataset from a csv file.

```python
data = pandas.read_csv('./preprocessed_titanic_data.csv')
data.head()
```

### Features-labels split and train-validation split

```python
features = data.drop(["Survived"], axis=1)
labels = data["Survived"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
# remark: we fix random_state the end, to make sure we always get the same split
features_train, features_validation_test, labels_train, labels_validation_test = train_test_split(
    features, labels, test_size=0.4, random_state=100)
```

```python
features_validation, features_test, labels_validation, labels_test = train_test_split(
    features_validation_test, labels_validation_test, test_size=0.5, random_state=100)
```

```python
print(len(features_train))
print(len(features_validation))
print(len(features_test))
print(len(labels_train))
print(len(labels_validation))
print(len(labels_test))
```

### Training model on our dataset


```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(features_train, labels_train)
```

### Model Evaluation

#### Accuracy

```python
print("Scores of the model")
# use model score to print model accuracy
print(lr_model.score(features_test,labels_test))
```

#### False Positives, False Negatives and Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
# use sklearn confusion_matrix, ConfusionMatrixDisplay to plot the confusion matrix
matrix1_test = confusion_matrix(lr_model.predict(features_test),labels_test)
matrix1_test
```

```python
matrix1_train = confusion_matrix(lr_model.predict(features_train),labels_train)
matrix1_train
```

```python
a=ConfusionMatrixDisplay(matrix1_test)
a.plot()
plt.show()
```

```python
matrix2_train = ConfusionMatrixDisplay(matrix1_train)
matrix2_train.plot()
plt.show()
```

#### Recall, Precision and F1-score

```python
from sklearn.metrics import f1_score,recall_score,precision_score
# use sklearn f1_score,recall_score,precision_score to print the f1_score,recall_score,precision_score
# don't use average parameter
print("f1 score for train  ",f1_score(lr_model.predict(features_train),labels_train))
print("f1 score for test  ",f1_score(lr_model.predict(features_test),labels_test))
print("#######################################")
print("recall_score for train  ",recall_score(lr_model.predict(features_train),labels_train))
print("recall_score for test  ",recall_score(lr_model.predict(features_test),labels_test))
print("#######################################")
print("precision_score for train  ",precision_score(lr_model.predict(features_train),labels_train))
print("precision_score for test  ",precision_score(lr_model.predict(features_test),labels_test))
```

### ROC and AUC

```python
from sklearn.metrics import RocCurveDisplay,auc,roc_curve
# use sklearn RocCurveDisplay to plot the ROC and AUC
fpr, tpr, thresholds = roc_curve(lr_model.predict(features_test),labels_test)
roc_auc = auc(fpr, tpr)
display = RocCurveDisplay(fpr=fpr, tpr=tpr, roc_auc=roc_auc)
```

```python
display.plot()
plt.show()
```
