## ما هو DeepSeek؟

**DeepSeek** هو نظام ذكاء اصطناعي متقدم يعتمد على نماذج لغوية كبيرة (Large Language Models - LLMs). يتميز DeepSeek بقدرته على فهم النصوص المعقدة وتوليدها بشكل طبيعي، مما يجعله خيارًا مثاليًا لبناء مساعدين ذكاء اصطناعي. يمكن استخدام DeepSeek لتطوير بوتات الدردشة، المساعدات الشخصية، وأدوات الدعم الفني.

### المرحلة 3: كتابة الكود

#### الخطوة 3.1: إنشاء ملف Python للتفاعل مع DeepSeek
قم بإنشاء ملف باسم deepseek_assistant.py وإضافة الكود التالي:
   ```bash
   import requests
from flask import Flask, request, jsonify

# إعدادات DeepSeek API
API_URL = "http://localhost:8000/api/v1/generate"
API_KEY = "YOUR_DEEPSEEK_API_KEY"   # ده api key الي هيجي من موقع ديب سيك لزم يتغير

app = Flask(__name__)

def ask_deepseek(prompt):
    headers = {
        "Authorization": f"Bearer {API_KEY}"
    }
    data = {
        "prompt": prompt,
        "max_tokens": 100
    }
    response = requests.post(API_URL, headers=headers, json=data)
    return response.json().get("response", "عذراً، لم أتمكن من فهم سؤالك.")

@app.route('/ask', methods=['POST'])
def handle_request():
    user_message = request.json.get('message', '')
    if not user_message:
        return jsonify({"error": "لا يوجد نص في الرسالة."}), 400
    assistant_response = ask_deepseek(user_message)
    return jsonify({"response": assistant_response})

if __name__ == '__main__':
    app.run(debug=True, port=5000)
   ```

## شرح الكود:

- **طلب بيانات من DeepSeek**: يتم إرسال النص الوارد (**prompt**) إلى **DeepSeek** عبر طلب **POST**.
- **إنشاء واجهة خدمة (API)**: تم استخدام **Flask** لإنشاء نقطة نهاية `/ask` تتلقى الرسائل الواردة وترسلها إلى **DeepSeek**.
- **إرجاع الردود**: يتم إرسال الردود إلى المستخدم في شكل **JSON**.

## الخاتمة
باستخدام **DeepSeek**، يمكنك بناء **مساعد ذكاء اصطناعي** قوي ومفيد بأقل جهد ممكن. سواء كنت ترغب في تحسين **تجربة العملاء** أو إنشاء **بوت دردشة مخصص**، فإن **DeepSeek** يقدم **أدوات مرنة وفعالة** لتحقيق ذلك.  
مع بعض الخطوات البسيطة والمتابعة المستمرة، يمكنك تطوير مساعد ذكي يلبي احتياجاتك **ويتجاوز توقعات المستخدمين**.


````python
import requests

url = "https://ghghthtj.app.n8n.cloud/webhook-test/249fe705-07d9-4f66-a0f2-de3c60b0594b"

def chat_bot():
    while True:
        user_input = input("اسألني عن كتب البرمجة أو اكتب 'خروج' لإنهاء الدردشة: ").strip()
        
        if user_input.lower() == "خروج":
            print("👋 وداعًا! نتمنى لك قراءة ممتعة 📚")
            break

        response = requests.post(url, json={"question": user_input})
        
        if response.status_code == 200:
            data = response.json()
            answer = data.get("answer", "لم أحصل على إجابة مناسبة.")
            print(f"🤖: {answer}")
        else:
            print("❌ حدث خطأ أثناء الاتصال بالخادم.")

chat_bot()
````
